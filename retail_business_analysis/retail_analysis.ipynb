```python
# retail analysis using jupyter notebook
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"dataset\superstore.csv", encoding='latin1')

df.columns = [c.replace(" ", "_").replace("-", "_") for c in df.columns]
df.to_csv("superstore_clean.csv", index=False)
print(df.columns)
```

```python
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.describe())
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'])
```

```python
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
```

```python
print(df['Sales'].sum())
```

```python
print(df['Profit'].sum())
```

```python
print(df.groupby('Category')['Sales'].sum().sort_values(ascending=False)) 
```

```python
print(df.groupby('Category')['Profit'].sum().sort_values(ascending=False)) 
```

```python
print(df.groupby('Region')['Sales'].sum().sort_values(ascending=False))
```

```python
print(df.groupby('Region')['Profit'].sum().sort_values(ascending=False))
```

```python
print(df.groupby('Discount')['Profit'].mean())
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
category_profit.plot(kind='bar')
plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
region_sales.plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
region_profit.plot(kind='bar')
plt.title('Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
discount_profit = df.groupby('Discount')['Profit'].mean()
discount_profit.plot(kind='line', marker='o')
plt.title('Discount vs Average Profit')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.grid(True)
plt.show()
```

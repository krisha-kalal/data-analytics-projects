```python
# retail analysis using jupyter notebook
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"dataset\superstore.csv", encoding='latin1')

df.columns = [c.replace(" ", "_").replace("-", "_") for c in df.columns]
df.to_csv("superstore_clean.csv", index=False)
print(df.columns)
```

```python
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.describe())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
```

```python
print(df['Sales'].sum())
```

```python
print(df['Profit'].sum())
```

```python
print(df.groupby('Category')['Sales'].sum().sort_values(ascending=False)) 
```

```python
print(df.groupby('Category')['Profit'].sum().sort_values(ascending=False)) 
```

```python
print(df.groupby('Region')['Sales'].sum().sort_values(ascending=False))
```

```python
print(df.groupby('Region')['Profit'].sum().sort_values(ascending=False))
```

```python
print(df.groupby('Discount')['Profit'].mean())
```
